### SOS, HELP
* Si necesitamos saber que hace alguna función o método de python, el nos da una una ayuda muy bien documentada con **Help**

```python
help(str.upper) #Preguntamos por la ayuda de la función upper
help(str) #Preguntamos por la ayuda de la clase str
```

Calculate Maximum

```python
student_grades = [9.1, 8.8, 7.5]
max_value = max(student_grades)
print(max_value)
```

Count Values (E)

```python
student_grades = [9.1, 8.8, 10.0, 7.7, 6.8, 8.0, 10.0, 8.1, 10.0, 9.9]
print(student_grades.count(10.0))
```
